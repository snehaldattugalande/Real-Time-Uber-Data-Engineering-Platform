```python
import pandas as pd

files = [
{"file":"map_cities"},
{"file":"map_cancellation_reasons"},
{"file":"map_payment_methods"},
{"file":"map_ride_statuses"},
{"file":"map_vehicle_makes"},
{"file":"map_vehicle_types"}
]

for file in files:

    #url = f"https://dl1uberprojectdev.blob.core.windows.net/raw/ingestion/{file['file']}.json?sp=r&st=2026-05-23T17:11:30Z&se=2026-06-06T01:26:30Z&spr=https&sv=2026-02-06&sr=c&sig=TFOu5dhfDonTycf2jn0JO893gUkbHdEFL9%2FuYB3Tv4Y%3D"
    url = f"https://dl1uberprojectdev.blob.core.windows.net/raw/ingestion/{file['file']}.json?sp=r&st=2026-05-28T13:28:26Z&se=2026-05-28T21:43:26Z&spr=https&sv=2026-02-06&sr=d&sig=cXn%2FVUKtc0kMlbO4zgZx2PT%2BusnEuzzXlqIL88qzO20%3D&sdd=1"
    
    df = pd.read_json(url)
    df_spark = spark.createDataFrame(df)
    
    # Writing Data To the Bronze Layer
    df_spark.write.format("delta")\
            .mode("overwrite")\
            .option("overwriteSchema", "true")\
            .saveAsTable(f"uber.bronze.{file['file']}")
```
